```python
df_cleaned = df.dropna()

df_cleaned.rename(columns={
    'MolecularWeight': 'Molecular_Weight',
    'XLogP': 'LogP',
    'CanonicalSMILES': 'SMILES',
    'InChIKey': 'InChI_Key'
}, inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(data=df_cleaned, x='Molecular_Weight', y='LogP')
plt.title('Molecular Weight vs LogP')
plt.xlabel('Molecular Weight')
plt.ylabel('LogP')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=df_cleaned, x='InChI_Key', y='LogP')
plt.title('LogP Values of Compounds')
plt.xlabel('Compound (InChIKey)')
plt.ylabel('LogP')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# Explain how trends in these molecular properties can support solvent selection in drug development.

Such molecular characteristics as molecular weight and LogP are important in the choice of solvents for drug development. Molecular weight influences the ease by which a compound diffuses in biological membranes. Highly molecular compounds can find it difficult to dissolve or be absorbed effectively, and thus solvents need to be used to facilitate their solubility or decrease their viscosity.

LogP reveals the lipophilicity of a compound, with its affinity for fat or aqueous environments. A high LogP indicates that the compound is more lipophilic, that is, more soluble in non-polar solvents. Low LogP refers to hydrophilic behavior and polar solvents such as water or ethanol would be more suitable.

By examining trends in these properties across a series of compounds, formulators can select solvents that most effectively balance solubility, stability, and bioavailability. This makes the drug both effective and convenient to administer.
